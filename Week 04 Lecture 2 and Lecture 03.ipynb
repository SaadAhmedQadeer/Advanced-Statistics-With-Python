## What is Machine Learning?

# Part 2: Regression Analysis — A Complete Introduction

---

## What is Regression Analysis?

## Definition

Regression analysis examines relationships between:
- **Independent variables** (predictors, features, inputs)
- **Dependent variable** (outcome, target, response)

## Core Goals
1. **Understand:** How input changes relate to outcome changes
2. **Predict:** Estimate outcomes for new data
3. **Infer:** Determine if relationships are statistically meaningful or due to chance

## Historical Context

- **Over 200 years:** Regression in use for statistics
- **1880s:** Francis Galton studied parent-child heights
- **Key Finding:** Children of very tall parents tend toward average — "regression to the mean"
- **Name Origin:** Statistical method named after this observation

---

## Why Regression is So Powerful

## Regression as a Reasoning Tool

Regression answers causal and inferential questions:
- **Ceteris paribus:** Holding all else constant, what is the effect of each variable?
- **Statistical significance:** Is an effect real or due to random variation?
- **Explanatory power:** How much variation do our variables explain?
- **Missing factors:** Are important predictors absent?

This goes beyond pattern-matching — it enables causal reasoning

These are causal and inferential questions — not just pattern-matching. This is what separates regression from many purely predictive algorithms, which can tell you what will happen but not why.

---

## Types of Regression

Regression is a family of methods, not a single algorithm. Here are the main types and when you would use each one:

## Types of Regression

**Simple Linear Regression**
- One input variable → continuous output
- Straight-line relationship
- Example: salary from experience

**Multiple Linear Regression**
- Two+ input variables → continuous output
- Example: insurance cost from age, BMI, smoking, dependents

**Polynomial Regression**
- Curved, non-linear relationships
- Inputs raised to powers
- Example: crop yield with fertilizer (increases then decreases)

**Ridge & Lasso Regression**
- Regularized versions with complexity penalties
- Prevent overfitting with many features

**Logistic Regression**
- Binary classification (despite name)
- Outputs probability between 0 and 1
- Example: disease/healthy prediction

**Poisson Regression**
- Count outcomes
- Example: accidents per month, customer complaints

Today we will focus primarily on simple and multiple linear regression, with a look at polynomial regression. These three cover the vast majority of real-world regression use cases.

---

## The Mathematics Behind Linear Regression

## Simple Linear Regression: The Model

```
Y = beta_0 + beta_1 * X + epsilon
```

**Components:**
- **Y:** Outcome we predict (dependent variable)
- **X:** Input variable (independent variable)
- **beta_0:** Intercept (predicted Y when X = 0)
- **beta_1:** Slope (expected Y change per one-unit X increase)
- **epsilon:** Error term (unexplained variation)

## Finding the Best Fit: Ordinary Least Squares (OLS)

**Method:**
- Minimize Sum of Squared Residuals (SSR)
- SSR = total squared difference between actual and predicted Y

**Why Square the Errors?**
1. Prevents positive and negative errors from canceling
2. Creates smooth, differentiable function for analytical optimization
3. Penalizes large errors more heavily

## Calculating the Slope

```
beta_1 = Covariance(X,Y) / Variance(X)
beta_1 = sum((Xi - X_mean) * (Yi - Y_mean)) / sum((Xi - X_mean)^2)
```

**Interpretation:**
- Measures how much X and Y move together
- Relative to how much X varies on its own

In multiple linear regression, the model becomes:

```
Y = beta_0 + beta_1*X1 + beta_2*X2 + ... + beta_p*Xp + epsilon
```

The same OLS principle applies, but now we are finding coefficients for each of the p input variables simultaneously.

---

## The Five Assumptions of Linear Regression

## Why Assumptions Matter

These conditions are **not arbitrary**:
- Mathematical requirements for unbiased OLS estimates
- Necessary for valid significance tests
- Violations lead to incorrect p-values and confidence intervals

**1. Linearity**
- Relationship between inputs and outcome must be linear
- Curved relationships → systematically wrong predictions
- Checked with: scatter plots, residual plots

**2. Independence of Errors**
- Each observation must be independent
- Errors must not be correlated
- Violated when: time series data, repeated measurements from same person/group

**3. Homoscedasticity (Equal Variance)**
- Residual spread constant across all predictor levels
- Heteroscedasticity: residuals fan out → WRONG standard errors → WRONG significance tests

**4. Normality of Errors**
- Small samples: residuals should be approximately normal
- Needed for valid t-tests and confidence intervals
- Large samples (100+): Central Limit Theorem provides protection

**5. No Perfect Multicollinearity**
- Input variables should not be perfectly correlated
- Perfect correlation: Model cannot distinguish separate effects
- High correlation: Inflates standard errors → hard to identify important variables

We will check all of these assumptions in our project today.

# Step 4: Data Preprocessing

## Data Preprocessing

**Requirement:** Regression needs all numerical inputs

### Encoding Categorical Variables

**Method:** One-hot encoding (dummy variable encoding)
- Replace categorical column with binary (0/1) columns
- One column per category
- Example: region (4 categories) → 4 binary columns

**Dummy Variable Trap:**
- Drop one category per variable (reference category)
- All effects interpreted relative to reference
- Prevent perfect multicollinearity
- Implementation: `drop_first=True`

### Feature Scaling

**OLS Regression:**
- Mathematically does not require scaling
- Coefficients stay correct

**Why Scale?**
- Coefficients become comparable in effect size
- Both in same standardized units

**Our Approach:**
- Keep original scale for interpretability
- Note scaling implications when needed

````python
# Create a copy so we always have the original data available
df_model = df.copy()

# One-hot encode the categorical variables
# drop_first=True avoids the dummy variable trap
df_encoded = pd.get_dummies(df_model,
                             columns=['sex', 'smoker', 'region'],
                             drop_first=True)

print('Columns after encoding:')
print(list(df_encoded.columns))
print()
print('Shape after encoding:', df_encoded.shape)
print()
print('First 3 rows of the encoded dataset:')
df_encoded.head(3)
````

## Why Train-Test Split?

**Problem:** Training and evaluating on same data = measuring memorization, not prediction

**Overfitting:**
- Model learns training data noise
- Poor performance on new data
- Good training score ≠ good model

**Solution:** Hold-out test set
- 80% training, 20% testing
- Test data never seen during training
- Honest estimate of real-world performance

**Reproducibility:**
- `random_state=42` ensures same split every run

## Reading the statsmodels Summary

**R-squared (R²):**
- Proportion of variation explained
- Example: R² = 0.09 means 9% explained; 91% from other sources

**Coefficient Table (The Heart):**
- **coef:** Estimated effect size (beta_1)
  - Example: age coef = 257 → $257 more per year
- **std err:** Uncertainty in estimate
  - Smaller = more precise
- **t:** t-statistic (coef / std_err)
  - How many std errors away from zero
- **P>|t|:** P-value
  - P < 0.05 = statistically significant
  - P ≈ 0.000 = highly significant
- **[0.025 0.975]:** 95% confidence interval
  - Range of likely true values

**F-statistic:**
- Tests overall model significance
- Does ANY variable help explain outcome?

**Critical Reminder:**
- Statistical significance ≠ practical significance
- Large p-value (5x) can still matter if coefficient is large
- Look at coefficient size AND p-value, not just p-value

# Step 6: Multiple Linear Regression

Multiple linear regression extends the simple case to include multiple input variables simultaneously. The model is:

Each coefficient now represents the **partial effect** of that variable — the effect of a one-unit change in that variable holding all other variables constant. For example, the coefficient on BMI in the multiple regression will tell us: comparing two people who are the same age, the same sex, smoke the same amount, live in the same region, and have the same number of children — how much do their charges differ if one has a BMI one unit higher? This controls for confounding in a way that simple correlation cannot.

````python
# Evaluate the model on both training and test data using sklearn

# Fit sklearn model (for prediction metrics)
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

# Predictions on both sets
y_train_pred = lr_model.predict(X_train)
y_test_pred = lr_model.predict(X_test)

# Calculate metrics
def evaluate_model(y_true, y_pred, set_name):
    r2 = r2_score(y_true, y_pred)
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    print(f'--- {set_name} ---')
    print(f'  R-squared (R2):                  {r2:.4f}')
    print(f'  Mean Absolute Error (MAE):       ${mae:,.2f}')
    print(f'  Root Mean Squared Error (RMSE):  ${rmse:,.2f}')
    print(f'  Mean Abs Percentage Error (MAPE): {mape:.2f}%')
    print()

print('Model Performance Metrics')
print('=' * 45)
evaluate_model(y_train, y_train_pred, 'Training Set')
evaluate_model(y_test, y_test_pred, 'Test Set')
````

**Understanding the evaluation metrics:**

This is one of the most important sections of any regression analysis. These four metrics each tell us something different about how the model performs.

**R-squared (R2):** The proportion of variance in charges explained by the model. An R2 of 0.75 means the model explains 75% of the variation in charges. The remaining 25% is either noise or comes from variables we do not have. For this kind of real-world data with genuine uncertainty, an R2 in the range of 0.70 to 0.80 would be considered good.

**Mean Absolute Error (MAE):** The average absolute difference between predicted and actual charges, in dollars. This is easy to interpret — if the MAE is $4,000, then on average your predictions are off by $4,000. MAE treats all errors equally regardless of their size.

**Root Mean Squared Error (RMSE):** The square root of the average squared errors. Unlike MAE, RMSE penalizes large errors more heavily because errors are squared before averaging. An RMSE significantly larger than the MAE indicates the model is making some large prediction errors, even if the average error is moderate.

**Mean Absolute Percentage Error (MAPE):** The average error expressed as a percentage of the actual value. This is scale-independent — a MAPE of 30% means predictions are typically off by 30% of the true value. It can be misleading when some true values are near zero.

**Comparing train and test performance:** If the test metrics are much worse than the training metrics, the model is overfit — it learned patterns specific to the training data that do not generalize. Linear regression is relatively resistant to overfitting, especially with a dataset this size.

**Which metric should you optimize for?** It depends on the business context. For insurance pricing, large errors are particularly costly — being off by $20,000 on a single policyholder is far worse than being off by $2,000 on ten policyholders, even though the total error might be similar by MAE. In that case, RMSE (which penalizes large errors more) is more appropriate.

# Step 7: Checking the Regression Assumptions

We discussed the five assumptions of linear regression earlier. Now we will check them formally using diagnostic plots and statistical tests. This is a critical step that is often skipped in introductory courses, but it is what separates a careful analysis from a careless one.

If assumptions are violated, the coefficient estimates may still be valid but the standard errors — and therefore the p-values and confidence intervals — may be wrong. This means you could be making incorrect conclusions about which variables are significant.

````python
# Comprehensive regression diagnostics
# Using the multiple regression model fit with statsmodels

residuals_full = multiple_model.resid
fitted_full = multiple_model.fittedvalues

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Residuals vs. Fitted (check linearity and homoscedasticity)
axes[0,0].scatter(fitted_full, residuals_full, alpha=0.35, color='steelblue', s=15)
axes[0,0].axhline(0, color='red', linestyle='--', linewidth=1.5)
# Add a LOWESS smoother to help see patterns
from statsmodels.nonparametric.smoothers_lowess import lowess
smoothed = lowess(residuals_full, fitted_full, frac=0.3)
axes[0,0].plot(smoothed[:,0], smoothed[:,1], color='orange', linewidth=2, label='LOWESS trend')
axes[0,0].set_xlabel('Fitted Values')
axes[0,0].set_ylabel('Residuals')
axes[0,0].set_title('1. Residuals vs. Fitted Values\n(Check linearity and equal variance)', fontsize=11)
axes[0,0].legend(fontsize=9)

# 2. Normal Q-Q Plot (check normality of residuals)
from scipy import stats as scipy_stats
standardized_residuals = (residuals_full - residuals_full.mean()) / residuals_full.std()
sm.qqplot(residuals_full, line='s', ax=axes[0,1], alpha=0.5)
axes[0,1].set_title('2. Normal Q-Q Plot\n(Check normality of residuals)', fontsize=11)

# 3. Scale-Location Plot (check homoscedasticity)
sqrt_abs_resid = np.sqrt(np.abs(standardized_residuals))
axes[1,0].scatter(fitted_full, sqrt_abs_resid, alpha=0.35, color='steelblue', s=15)
smoothed2 = lowess(sqrt_abs_resid, fitted_full, frac=0.3)
axes[1,0].plot(smoothed2[:,0], smoothed2[:,1], color='orange', linewidth=2)
axes[1,0].set_xlabel('Fitted Values')
axes[1,0].set_ylabel('Sqrt(|Standardized Residuals|)')
axes[1,0].set_title('3. Scale-Location Plot\n(Check homoscedasticity)', fontsize=11)

# 4. Residuals distribution
axes[1,1].hist(residuals_full, bins=50, color='steelblue', edgecolor='white', alpha=0.85, density=True)
x_range = np.linspace(residuals_full.min(), residuals_full.max(), 200)
normal_curve = scipy_stats.norm.pdf(x_range, residuals_full.mean(), residuals_full.std())
axes[1,1].plot(x_range, normal_curve, 'r-', linewidth=2, label='Normal distribution')
axes[1,1].set_xlabel('Residuals')
axes[1,1].set_ylabel('Density')
axes[1,1].set_title('4. Residuals Distribution\n(Compared to normal curve)', fontsize=11)
axes[1,1].legend()

plt.suptitle('Regression Assumption Diagnostics', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()

# Statistical tests for normality
stat_shapiro, p_shapiro = shapiro(residuals_full[:500])  # Shapiro-Wilk uses first 500 for performance
stat_jb, p_jb = jarque_bera(residuals_full)

print('Normality Tests on Residuals:')
print(f'  Shapiro-Wilk test:   statistic={stat_shapiro:.4f}, p-value={p_shapiro:.6f}')
print(f'  Jarque-Bera test:    statistic={stat_jb:.2f}, p-value={p_jb:.6f}')
print()
if p_jb < 0.05:
    print('  Result: Residuals are NOT normally distributed (p < 0.05).')
    print('  This is expected given the right-skewed nature of insurance charges.')
    print('  With n=1338 observations, the CLT helps, but a log transformation')
    print('  of the target variable would improve this.')
else:
    print('  Result: Cannot reject normality of residuals (p >= 0.05).')
````

**Interpreting the diagnostic plots:**

**Plot 1 — Residuals vs. Fitted:** Ideally, points should be randomly scattered around the horizontal zero line with no pattern. If the orange LOWESS line curves away from zero, there is evidence of non-linearity. If the spread of residuals changes dramatically across the x-axis, there is heteroscedasticity. The two-band pattern we discussed (smokers vs. non-smokers) will still be visible here.

**Plot 2 — Q-Q Plot:** This is a powerful visual test for normality. The x-axis shows the theoretical quantiles of a normal distribution. The y-axis shows the actual quantiles of our residuals. If the residuals were perfectly normal, all points would fall on the 45-degree reference line. Deviations at the ends (tails) indicate heavier or lighter tails than normal. Deviations in a consistent curve indicate systematic skewness.

**Plot 3 — Scale-Location:** Similar to Plot 1 but using the square root of absolute standardized residuals. A horizontal, flat orange line would indicate constant variance. If the line slopes upward, larger fitted values have more variable residuals.

**Plot 4 — Residuals distribution vs. normal:** A direct comparison of the residuals histogram with what a normal distribution of the same mean and standard deviation would look like.

**The normality test results:** The p-value for the Jarque-Bera test is likely very small (near zero), meaning we reject normality. This is not surprising given what we know about the skewed distribution of charges. For a dataset of 1,338 observations, this violation is not catastrophic — the Central Limit Theorem provides some protection — but it does mean our p-values and confidence intervals should be interpreted with some caution. The proper fix is to either model log(charges) or use heteroscedasticity-robust standard errors.

````python
# Check for multicollinearity using Variance Inflation Factor (VIF)
# VIF measures how much the variance of a coefficient is inflated due to
# correlation with other predictors.
# Rule of thumb: VIF > 5 is concerning, VIF > 10 is a serious problem.

# Convert boolean columns to integers (0s and 1s) for statsmodels compatibility
X_train_numeric = X_train.astype(int)

X_vif = sm.add_constant(X_train_numeric) # Use the numeric version for VIF calculation
vif_data = pd.DataFrame()
vif_data['Feature'] = X_train_numeric.columns # Use numeric version for column names
vif_data['VIF'] = [variance_inflation_factor(X_vif.values, i) # Pass X_vif.values (with constant) and use the loop for columns including the constant
                    for i in range(X_vif.shape[1])]
vif_data = vif_data.drop(vif_data.index[0]) # Drop the 'const' row from VIF results
vif_data = vif_data.sort_values('VIF', ascending=False)

print('Variance Inflation Factors:')
print(vif_data.to_string(index=False))
print()
print('VIF interpretation:')
print('  VIF = 1: No correlation with other predictors')
print('  VIF 1-5: Moderate, generally acceptable')
print('  VIF > 5: Concerning, consider removing or combining variables')
print('  VIF > 10: Serious multicollinearity, action required')
````

**Understanding VIF and why multicollinearity matters:**

Multicollinearity occurs when predictor variables are correlated with each other. When this happens, it becomes difficult for the model to isolate the individual effect of each variable.

Imagine trying to separate the effects of height and weight on athletic performance — they are correlated, so it is hard to say "holding weight constant, what is the effect of height?" because in reality, taller people also tend to be heavier.

A VIF of 10 for a variable means that its standard error is sqrt(10) ≈ 3.16 times larger than it would be if that variable were uncorrelated with all others. This inflated standard error makes the t-statistic smaller, potentially making a truly important variable appear statistically insignificant.

For this dataset, VIF values should all be relatively low (under 5), since the predictors are not strongly correlated with each other. This is one of the cleaner datasets for demonstrating regression without multicollinearity concerns.

# Step 8: Polynomial Regression

Sometimes the relationship between a predictor and the outcome is not linear — it curves. Polynomial regression extends linear regression to handle these non-linear relationships by adding powers of the input variables.

For example, if we suspect that the relationship between BMI and charges is not a straight line but a curve (perhaps costs accelerate dramatically above BMI=30), we might model:

```
charges = beta_0 + beta_1*bmi + beta_2*bmi^2 + epsilon
```

Even though this model contains bmi-squared, it is still linear regression in the mathematical sense — linear in the parameters (beta coefficients). We are just creating a new variable called bmi-squared and treating it as any other predictor.

**Why not always use polynomial regression?**

Higher-degree polynomials can overfit badly. A degree-10 polynomial can pass through every single data point perfectly — but it will perform terribly on new data because it has learned the noise, not the signal. Cross-validation is essential for choosing the right degree.

**When to use polynomial regression:**
- When your residual plots show a clear curved pattern
- When you have domain knowledge suggesting a non-linear relationship
- As a starting point before trying more complex non-linear models

Let us explore polynomial regression specifically for BMI vs. charges.

````python
# Polynomial Regression: Comparing degree 1, 2, and 3 for age vs. charges
# We will use sklearn's PolynomialFeatures within a Pipeline

# Using age as the single predictor for clear visualization
X_age = df[['age']].values
y_charges = df['charges'].values

# Split
X_age_train, X_age_test, y_age_train, y_age_test = train_test_split(
    X_age, y_charges, test_size=0.2, random_state=42
)

degrees = [1, 2, 3]
colors_poly = ['#e74c3c', '#2ecc71', '#9b59b6']
results = []

fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(X_age_train, y_age_train, alpha=0.25, color='steelblue', s=15, label='Training data')
ax.scatter(X_age_test, y_age_test, alpha=0.25, color='orange', s=15, label='Test data', marker='^')

age_range_plot = np.linspace(df['age'].min(), df['age'].max(), 300).reshape(-1, 1)

for deg, color in zip(degrees, colors_poly):
    # Pipeline: transform features to polynomial, then fit linear regression
    poly_pipeline = Pipeline([
        ('poly', PolynomialFeatures(degree=deg, include_bias=True)),
        ('linear', LinearRegression())
    ])
    poly_pipeline.fit(X_age_train, y_age_train)

    # Predict on test set
    y_pred_test = poly_pipeline.predict(X_age_test)
    train_r2 = poly_pipeline.score(X_age_train, y_age_train)
    test_r2 = poly_pipeline.score(X_age_test, y_age_test)
    rmse_test = np.sqrt(mean_squared_error(y_age_test, y_pred_test))

    results.append({'Degree': deg, 'Train R2': train_r2, 'Test R2': test_r2, 'Test RMSE': rmse_test})

    # Plot the curve
    y_curve = poly_pipeline.predict(age_range_plot)
    ax.plot(age_range_plot, y_curve, color=color, linewidth=2.5,
             label=f'Degree {deg} (Test R2={test_r2:.3f})')

ax.set_xlabel('Age (years)')
ax.set_ylabel('Charges (USD)')
ax.set_title('Polynomial Regression: Age vs. Charges (Degree 1, 2, 3)')
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()

print('Model Comparison: Polynomial Degree')
print('=' * 55)
results_df = pd.DataFrame(results)
results_df['Train R2'] = results_df['Train R2'].map('{:.4f}'.format)
results_df['Test R2'] = results_df['Test R2'].map('{:.4f}'.format)
results_df['Test RMSE'] = results_df['Test RMSE'].map('${:,.0f}'.format)
print(results_df.to_string(index=False))
````

**What to look for in the polynomial comparison:**

Typically, you will see that degree 2 and degree 3 have slightly higher Train R2 than degree 1 — they fit the training data better. But the Test R2 (performance on data the model never saw) may not improve much, or could even decrease. This demonstrates the risk of overfitting.

For age vs. charges, the relationship is likely close to linear, so the higher-degree polynomials are not adding real information — they are adding complexity without benefit.

**The general principle:** Always evaluate model complexity on out-of-sample data (the test set). Train performance only tells you how well the model memorized; test performance tells you how well it learned.

# Step 9: Capturing Interaction Effects

Earlier, we observed something important in the scatter plots: the effect of BMI on charges seemed to depend on whether the person smokes. Specifically, smokers with high BMI seemed to have dramatically higher costs than both smokers with low BMI and non-smokers with high BMI.

This is called an **interaction effect**. Two variables interact when the effect of one on the outcome depends on the value of the other.

In a standard multiple regression model, we assume that the effects of all variables are **additive and independent**. If we include both BMI and smoker status, the model assumes that the BMI effect is the same for smokers and non-smokers, and the smoking effect is the same at all BMI levels. This assumption is wrong here.

To capture the interaction, we create a new variable: **BMI multiplied by smoker status** (which is 0 or 1). This interaction term allows the BMI slope to be different for smokers than for non-smokers.

The model becomes:

```
charges = beta_0 + beta_1*bmi + beta_2*smoker + beta_3*(bmi * smoker) + other_terms + epsilon
```

For a non-smoker (smoker=0): charges = beta_0 + beta_1*bmi + other terms

For a smoker (smoker=1): charges = (beta_0 + beta_2) + (beta_1 + beta_3)*bmi + other terms

The interaction term (beta_3) tells us how much the BMI slope changes for smokers compared to non-smokers.

````python
# Build model with interaction term: bmi * smoker
# In statsmodels formula API, the colon (:) creates an interaction,
# and the star (*) creates both main effects AND the interaction

formula_interaction = 'charges ~ age + bmi * C(smoker) + children + C(sex) + C(region)'
interaction_model = smf.ols(formula_interaction, data=df).fit()

print('Model with BMI x Smoker Interaction Term')
print('=' * 50)
print(interaction_model.summary())
````

````python
# Compare base model vs. interaction model
base_r2 = multiple_model.rsquared_adj
interaction_r2 = interaction_model.rsquared_adj

print('Model Comparison:')
print(f'  Base multiple regression (Adjusted R2):       {base_r2:.4f}')
print(f'  Interaction model (Adjusted R2):              {interaction_r2:.4f}')
print(f'  Improvement from interaction term:            {interaction_r2 - base_r2:.4f}')
print()

# Visualize the interaction: separate regression lines for smokers vs. non-smokers
fig, ax = plt.subplots(figsize=(9, 6))

smoker_colors = {'yes': '#e74c3c', 'no': '#3498db'}
smoker_labels = {'yes': 'Smoker', 'no': 'Non-smoker'}

for smoker_val, group in df.groupby('smoker'):
    ax.scatter(group['bmi'], group['charges'],
               color=smoker_colors[smoker_val], alpha=0.3, s=15,
               label=f'{smoker_labels[smoker_val]} data')

# Plot the fitted interaction lines
bmi_range = np.linspace(df['bmi'].min(), df['bmi'].max(), 200)
mean_age = df['age'].mean()
mean_children = df['children'].mean()

for smoker_val, color in smoker_colors.items():
    pred_data = pd.DataFrame({
        'bmi': bmi_range,
        'age': mean_age,
        'children': mean_children,
        'sex': 'female',
        'smoker': smoker_val,
        'region': 'northeast'
    })
    predictions = interaction_model.predict(pred_data)
    ax.plot(bmi_range, predictions, color=color, linewidth=3,
             linestyle='-',
             label=f'{smoker_labels[smoker_val]} regression line')

ax.set_xlabel('BMI')
ax.set_ylabel('Predicted Charges (USD)')
ax.set_title('Interaction Effect: BMI x Smoker Status\n(at average age, children; female; northeast region)')
ax.legend(fontsize=9)
ax.axvline(30, color='gray', linestyle=':', linewidth=1.5, alpha=0.7)
ax.text(30.5, ax.get_ylim()[0] + 1000, 'BMI = 30\n(obese threshold)', fontsize=9, color='gray')
plt.tight_layout()
plt.show()
````

**What the interaction visualization reveals:**

This plot shows the fitted regression lines for smokers and non-smokers separately, as a function of BMI (holding all other variables at their mean or reference values).

For **non-smokers**, the line is relatively flat — increasing BMI has a modest positive effect on charges.

For **smokers**, the line has a much steeper positive slope — increasing BMI has a dramatically larger effect on charges when combined with smoking. The interaction coefficient captures this divergence.

The improvement in Adjusted R-squared from adding this single interaction term demonstrates that the interaction is not just theoretically interesting — it materially improves the model's ability to explain variation in charges.

**Practical implication:** An insurance company pricing model that ignores this interaction will underprice high-BMI smokers (who are actually much more costly) and overprice low-BMI smokers. The interaction model is more equitable and more accurate.

# Step 10: Cross-Validation — A More Honest Model Evaluation

When we split the data into training and test sets, the evaluation metrics depend on which specific observations happened to end up in the test set. If we got unlucky and all the high-cost outliers ended up in the test set, our metrics would look worse than they really are.

**K-fold cross-validation** addresses this by:
1. Dividing the data into K equal-sized groups (folds)
2. Training the model K times, each time using K-1 folds for training and the remaining fold for testing
3. Averaging the K test scores to get a more stable performance estimate

With K=5 (five-fold cross-validation), every data point is used for testing exactly once and for training four times. The final metric is the average of five independent evaluations.

This gives a more reliable estimate of how the model will perform on genuinely new data, and also gives us a sense of the **variance** of the performance estimate — how much the score changes depending on which data is in the test fold.

````python
# 5-fold cross-validation on the full dataset
from sklearn.model_selection import cross_validate, KFold

# Prepare the full encoded dataset for sklearn
X_full = df_encoded.drop('charges', axis=1)
y_full = df_encoded['charges']

kf = KFold(n_splits=5, shuffle=True, random_state=42)

cv_results = cross_validate(
    LinearRegression(), X_full, y_full, cv=kf,
    scoring=['r2', 'neg_mean_absolute_error', 'neg_root_mean_squared_error'],
    return_train_score=True
)

print('5-Fold Cross-Validation Results')
print('=' * 55)
print()
print(f'R-squared:')
print(f'  Train folds: {cv_results["train_r2"].mean():.4f} +/- {cv_results["train_r2"].std():.4f}')
print(f'  Test folds:  {cv_results["test_r2"].mean():.4f} +/- {cv_results["test_r2"].std():.4f}')
print()

mae_scores = -cv_results['test_neg_mean_absolute_error']
rmse_scores = -cv_results['test_neg_root_mean_squared_error']
print(f'Mean Absolute Error (MAE) across folds:')
for i, mae in enumerate(mae_scores):
    print(f'  Fold {i+1}: ${mae:,.0f}')
print(f'  Average: ${mae_scores.mean():,.0f} +/- ${mae_scores.std():,.0f}')
print()
print(f'RMSE across folds: ${rmse_scores.mean():,.0f} +/- ${rmse_scores.std():,.0f}')
````

**Interpreting cross-validation results:**

The key numbers to focus on are:

**The average test R-squared** gives us the most honest estimate of model performance. Compare this to the single test set R-squared we calculated earlier — they should be close, which would suggest our original train-test split was not giving us a misleadingly optimistic or pessimistic view.

**The standard deviation across folds** tells us how stable the performance is. A small standard deviation (say, 0.01 to 0.02) means the model performs consistently regardless of which subset of the data is used for testing. A large standard deviation would indicate the model's performance is sensitive to the specific training data, which could be a sign of overfitting or of genuine heterogeneity in the data.

**The difference between train and test scores** — if train R2 is 0.78 and test R2 is 0.77, the model generalizes well. If train is 0.95 and test is 0.70, we have a serious overfitting problem. For linear regression on this dataset, the gap should be small.

# Step 11: The Log-Transformed Model

Earlier, we noted that insurance charges are right-skewed and that log(charges) is more normally distributed. Let us now fit the model using log(charges) as the outcome and compare it to the original scale model.

**When to log-transform the outcome:**
- When the outcome has a right-skewed distribution (which violates the normality of errors assumption)
- When you expect multiplicative rather than additive effects (e.g., smoking increases costs by a percentage, not a fixed dollar amount)
- When the residuals from the linear model are clearly non-normal

**The interpretation changes:** With log(Y) as the outcome, coefficients are interpreted as the **percentage change** in Y for a one-unit change in X (approximately, for small coefficients). Specifically, a coefficient of 0.03 on age means a one-year increase in age is associated with approximately a 3% increase in charges. More precisely, the percentage change is exp(0.03) - 1 ≈ 3.04%.

````python
# Fit the log-transformed model
df_log = df.copy()
df_log['log_charges'] = np.log(df_log['charges'])

formula_log = 'log_charges ~ age + bmi + children + C(sex) + C(smoker) + C(region)'
log_model = smf.ols(formula_log, data=df_log).fit()

# Compare residual diagnostics
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Q-Q plot: original model residuals
sm.qqplot(multiple_model.resid, line='s', ax=axes[0], alpha=0.5)
axes[0].set_title(f'Q-Q Plot: Original Scale Model\n(R2={multiple_model.rsquared:.4f})', fontsize=11)

# Q-Q plot: log model residuals
sm.qqplot(log_model.resid, line='s', ax=axes[1], alpha=0.5)
axes[1].set_title(f'Q-Q Plot: Log-Transformed Model\n(R2={log_model.rsquared:.4f})', fontsize=11)

plt.suptitle('Residual Normality: Original vs. Log-Transformed Target', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()

print('Comparing the two models:')
print(f'  Original scale model - R2: {multiple_model.rsquared:.4f}, Adj-R2: {multiple_model.rsquared_adj:.4f}')
print(f'  Log-transformed model - R2: {log_model.rsquared:.4f}, Adj-R2: {log_model.rsquared_adj:.4f}')
print()
print('Note: R2 values are not directly comparable between the two models because')
print('they explain variance in different outcome variables (charges vs. log_charges).')
print()
print('Key log model coefficients and their interpretation:')
print()
for param, coef in log_model.params.items():
    if param != 'Intercept':
        pct_change = (np.exp(coef) - 1) * 100
        print(f'  {param}: coef={coef:.4f}, approx. {pct_change:+.1f}% change in charges per unit increase')
````

**Interpreting the log model coefficients:**

This is where the log model becomes particularly insightful. Instead of saying "smoking adds $23,000 to charges," we can now say "smoking is associated with approximately a 200% increase in charges" (the exact percentage depends on the coefficient). This percentage interpretation is often more meaningful because it accounts for the fact that a $23,000 increase means very different things for a person with $5,000 in base charges versus $50,000.

The Q-Q plots should show that the log model residuals are substantially closer to the normal reference line than the original model residuals. The deviation in the tails should be smaller, indicating that the log transformation has brought the residuals closer to the normality assumption.

In real-world insurance modeling, log-transformed cost models are very common precisely because the interpretations are more stable across the population.

# Step 12: Final Model Comparison and Summary

Let us bring everything together and compare the models we have built throughout this analysis. This section summarizes what we have learned and provides a framework for choosing between models in practice.

````python
# Final model comparison
# We compare the four main models we built using cross-validation

from sklearn.preprocessing import LabelEncoder

# Prepare encoded data for all models
X_final = df_encoded.drop('charges', axis=1)
y_original = df_encoded['charges']
y_log = np.log(df_encoded['charges'])

# Add interaction term: bmi * smoker_yes
X_interaction = X_final.copy()
X_interaction['bmi_x_smoker'] = X_final['bmi'] * X_final['smoker_yes']

kf = KFold(n_splits=5, shuffle=True, random_state=42)

model_configs = [
    ('Simple (age only)', X_final[['age']], y_original),
    ('Multiple Linear Regression', X_final, y_original),
    ('MLR + Interaction (BMI*Smoker)', X_interaction, y_original),
]

print('Final Model Comparison (5-fold CV)')
print('=' * 70)
print(f'{"Model":<40} {"Mean R2":>10} {"Std R2":>10} {"Mean MAE":>12}')
print('-' * 70)

for name, X_m, y_m in model_configs:
    cv = cross_validate(
        LinearRegression(), X_m, y_m, cv=kf,
        scoring=['r2', 'neg_mean_absolute_error'],
        return_train_score=False
    )
    mean_r2 = cv['test_r2'].mean()
    std_r2 = cv['test_r2'].std()
    mean_mae = -cv['test_neg_mean_absolute_error'].mean()
    print(f'{name:<40} {mean_r2:>10.4f} {std_r2:>10.4f} {mean_mae:>12,.0f}')

print('-' * 70)
print()
print('Note: MAE for log model is in log-dollar units and not comparable.')
print('For the log model, back-transform predictions with exp() before computing MAE.')
````

````python
# Standardized coefficient plot to compare effect sizes across variables
# Standardizing puts all coefficients on the same scale (standard deviations)
# so we can compare them directly

from sklearn.preprocessing import StandardScaler

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_final)
X_scaled_df = pd.DataFrame(X_scaled, columns=X_final.columns)

X_scaled_sm = sm.add_constant(X_scaled_df)
scaled_model = sm.OLS(y_original, X_scaled_sm).fit()

# Plot standardized coefficients
coef_std = pd.DataFrame({
    'Feature': X_final.columns,
    'Std_Coefficient': scaled_model.params[1:].values,
    'P_value': scaled_model.pvalues[1:].values
})
coef_std = coef_std.reindex(coef_std['Std_Coefficient'].abs().sort_values().index)

colors_std = ['#2ecc71' if p < 0.05 else '#e74c3c' for p in coef_std['P_value']]

fig, ax = plt.subplots(figsize=(9, 6))
bars = ax.barh(coef_std['Feature'], coef_std['Std_Coefficient'],
                color=colors_std, alpha=0.85, height=0.6)
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlabel('Standardized Coefficient (in standard deviations of charges per SD of predictor)', fontsize=10)
ax.set_title('Standardized Regression Coefficients\n(Green = significant p<0.05, allows comparison of effect sizes)', fontsize=11)

# Add value labels on the bars
for bar, val in zip(bars, coef_std['Std_Coefficient']):
    ax.text(val + (300 if val >= 0 else -300), bar.get_y() + bar.get_height()/2,
             f'{val:,.0f}', va='center', ha='left' if val >= 0 else 'right', fontsize=9)

plt.tight_layout()
plt.show()
````

## On Machine Learning vs. Deep Learning

Machine learning is the broader field of algorithms that learn from data. Deep learning is a specific subset using multi-layer neural networks. Deep learning excels at unstructured data like images, audio, and text, and becomes genuinely advantageous with very large datasets.

## On Regression Analysis

Linear regression is not just a prediction tool — it is an inferential framework. The coefficients are interpretable quantities with real-world meaning. The p-values and confidence intervals tell you what you can conclude about the population, not just the sample. The assumptions are not arbitrary rules — they are mathematical conditions that must be checked for the inference to be valid.

---

## On the Medical Insurance Project

The most important finding from our analysis was that **smoking is the dominant predictor of insurance charges** by a large margin — roughly 200% higher costs for smokers than non-smokers with comparable characteristics. The second most important finding was the **interaction between BMI and smoking** — high BMI compounds the effect of smoking dramatically. A model that misses these findings would produce incorrect predictions and misleading inferences.

This demonstrates why EDA is not optional. We discovered the smoking/non-smoker structure in the data through visualization before we built a single model, which then guided our modeling decisions.

---

## The Decision Framework for Regression in Practice

When approaching a regression problem, work through these questions in order:

1. Is the outcome truly continuous? (If not, consider logistic regression or other classification methods)
2. How is the outcome distributed? (Right-skewed outcomes often benefit from log transformation)
3. What does EDA reveal about individual variable relationships? (Look for non-linearity, clusters, outliers)
4. Are there likely interaction effects? (Check if relationships depend on other variables)
5. After fitting, always check the five assumptions — linearity, independence, homoscedasticity, normality, no perfect multicollinearity
6. Evaluate using held-out data, and prefer cross-validation for more stable estimates
7. Choose metrics appropriate for the business context

---

## What to Explore Next

The natural next steps beyond what we covered today are:

**Ridge and Lasso regression** — when you have many predictors or suspect overfitting, regularization methods add a penalty for large coefficients that shrinks the model and can perform automatic variable selection (Lasso).

**Time series regression** — when data points are ordered in time and the independence assumption is violated, specialized models like ARIMA and its variants handle temporal autocorrelation.

**Generalized Linear Models** — the extension of regression to non-Gaussian outcomes. Poisson regression for counts, logistic regression for binary outcomes, gamma regression for positively-skewed continuous outcomes.

**Robust regression** — methods that are resistant to outliers, using weighted least squares or alternative loss functions.
